```python
import pandas as pd
import random

# Categories
categories = [
    "Electronics", "Clothing", "Furniture", "Books",
    "Sports", "Beauty", "Groceries", "Toys",
    "Footwear", "Accessories"
]

category_df = pd.DataFrame({
    "Category_ID": range(1, 11),
    "Category_Name": categories
})

# Products - 1000 Records
products = []

for i in range(1, 1001):
    category_id = random.randint(1, 10)

    products.append({
        "Product_ID": i,
        "Product_Name": f"Product_{i}",
        "Category_ID": category_id,
        "Price": random.randint(100, 10000)
    })

product_df = pd.DataFrame(products)

print("Categories:", len(category_df))
print("Products:", len(product_df))

display(category_df.head())
display(product_df.head())
```

```python
import pandas as pd
import random

# Indian Cities and States
cities = {
    "Maharashtra": ["Mumbai", "Pune", "Nagpur", "Nashik"],
    "Karnataka": ["Bangalore", "Mysore", "Hubli"],
    "Gujarat": ["Ahmedabad", "Surat", "Vadodara"],
    "Delhi": ["New Delhi", "Delhi"],
    "Rajasthan": ["Jaipur", "Udaipur", "Jodhpur"],
    "Uttar Pradesh": ["Lucknow", "Kanpur", "Agra"],
    "Tamil Nadu": ["Chennai", "Coimbatore", "Madurai"],
    "Telangana": ["Hyderabad", "Warangal"]
}

first_names = [
    "Rahul", "Amit", "Priya", "Sneha", "Rohan",
    "Neha", "Anjali", "Vijay", "Karan", "Pooja"
]

last_names = [
    "Patil", "Sharma", "Verma", "Kumar", "Singh",
    "Joshi", "Deshmukh", "Shah", "Mehta", "Gupta"
]

customers = []

for i in range(1, 25001):
    state = random.choice(list(cities.keys()))
    city = random.choice(cities[state])

    customers.append({
        "Customer_ID": i,
        "Customer_Name": random.choice(first_names) + " " + random.choice(last_names),
        "City": city,
        "State": state
    })

customer_df = pd.DataFrame(customers)

print("Total Customers:", len(customer_df))
display(customer_df.head())
```

```python
import pandas as pd
import random

from datetime import datetime, timedelta

orders = []

start_date = datetime(2025, 1, 1)
end_date = datetime(2025, 12, 31)

date_range = (end_date - start_date).days

for i in range(1, 100001):

    order_date = start_date + timedelta(
        days=random.randint(0, date_range)
    )

    orders.append({
        "Order_ID": i,
        "Customer_ID": random.randint(1, 25000),
        "Order_Date": order_date,
        "Order_Status": random.choice(
            ["Delivered", "Shipped", "Cancelled", "Pending"]
        )
    })

orders_df = pd.DataFrame(orders)

print("Total Orders:", len(orders_df))
print("Start Date:", orders_df["Order_Date"].min())
print("End Date:", orders_df["Order_Date"].max())

display(orders_df.head())
```

```python
import pandas as pd

# Products Data
product_rows = [
    (int(r.Product_ID), str(r.Product_Name),
     int(r.Category_ID), float(r.Price))
    for r in product_df.itertuples(index=False)
]

# Orders Data
orders_df["Order_Date"] = pd.to_datetime(orders_df["Order_Date"])

order_rows = [
    (int(r.Order_ID), int(r.Customer_ID),
     r.Order_Date.to_pydatetime(), str(r.Order_Status))
    for r in orders_df.itertuples(index=False)
]

# Order Items Data
item_rows = [
    (int(r.Order_Item_ID), int(r.Order_ID),
     int(r.Product_ID), int(r.Quantity))
    for r in order_items_df.itertuples(index=False)
]

# Import Products
cursor.executemany("""
INSERT INTO Products
(Product_ID, Product_Name, Category_ID, Price)
VALUES (%s, %s, %s, %s)
""", product_rows)
conn.commit()
print("Products Imported:", len(product_rows))

# Import Orders
cursor.executemany("""
INSERT INTO Orders
(Order_ID, Customer_ID, Order_Date, Order_Status)
VALUES (%s, %s, %s, %s)
""", order_rows)
conn.commit()
print("Orders Imported:", len(order_rows))

# Import Order Items
for start in range(0, len(item_rows), 10000):
    cursor.executemany("""
    INSERT INTO Order_Items
    (Order_Item_ID, Order_ID, Product_ID, Quantity)
    VALUES (%s, %s, %s, %s)
    """, item_rows[start:start+10000])
    conn.commit()

print("Order Items Imported:", len(item_rows))

print("\nALL DATA IMPORT COMPLETED!")
```
